```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
```

#### Classify employees into experience levels based on TotalWorkingYears

```python
df["ExperienceLevel"] = df["TotalWorkingYears"].apply(
    lambda x: "Junior" if x < 5 else "Mid" if x <= 9 else "Senior"
)
```

#### Create a summary table showing the number of employees in each experience level

```python
experience_summary = df["ExperienceLevel"].value_counts().reset_index()
experience_summary.columns = ["Experience_Level", "Employee_Count"]
```

#### Sort the experience levels in a logical order

```python
experience_summary = experience_summary.set_index("Experience_Level").loc[
    ["Junior", "Mid", "Senior"]
].reset_index()
```
